# Appendix A: Formal Framework

# A.1 Notation

Let

$$d \in \{1,2\}$$

index datasets, where (d=1) denotes the original study and (d=2) denotes the replication study.

# A.2 Direct Replication

Direct replication targets the same estimand $\theta^*$ across studies:

$$\theta^{*(1)} \approx \theta^{*(2)}.$$

The corresponding exact equality condition is

$$\theta^{*(1)} = \theta^{*(2)},$$

# A.3 Conceptual Replication

Let $\theta^{(1)} \in \Theta_1$ and $\theta^{(2)} \in \Theta_2$ denote estimands arising from distinct operationalizations. Conceptual replication is defined through a mapping

$$\phi : \Theta_2 \rightarrow \Theta_1$$
such that

$$C(\theta^{(1)}) = C(\phi(\theta^{(2)})), $$

where $C(\cdot)$ denotes the theoretical claim supported by the estimand.

# A.4 Theory-Relevant Parameters

Let

$$\theta^{*}
:=
(\theta_{j_1}, \ldots, \theta_{j_r})$$

denote the subset of model parameters corresponding to the substantive claim of interest.

Replication at the parameter level may be expressed as

$$\theta^{*(1)}
\approx
\theta^{*(2)}.$$

The corresponding exact equality condition is

$$H_0:
\theta^{*(1)}
=
\theta^{*(2)}.$$

Approximate replication may be defined through substantively justified tolerances:

$$\left|
\hat\theta_j^{*(1)}
-
\hat\theta_j^{*(2)}
\right|
\le
\epsilon_j,
\qquad
\forall j,$$

where (\_j) denotes an acceptable discrepancy for parameter (j).

# A.5 Measurement and Structural Models

Let

$$ 
\mathbf{y}_i^{(g)} = \boldsymbol{\nu}^{(g)} + \boldsymbol{\Lambda}^{(g)} \boldsymbol{\eta}_i^{(g)} + \boldsymbol{\varepsilon}_i^{(g)}, $$

with

$$\boldsymbol{\varepsilon}_i^{(g)} \sim (0, \boldsymbol{\Theta}^{(g)}).$$

Where:

- $\mathbf{y}_i^{(g)}$: observed responses of individual $i$ in group $g$  
- $\nu^{(g)}$: item intercepts  
- $\eta_i^{(g)}$: latent variables for individual $i$  
- $\Lambda^{(g)}$: factor loadings linking latent variables to observed indicators  
- $\varepsilon_i^{(g)}$: residual errors
- $\Theta^{(g)}$: residual covariance matrix

Measurement Invariance is defined as

$$ \theta_M^{(g)} = \theta_M \quad \forall g ,$$

where

$$ \theta_M = \{\boldsymbol{\Lambda}, \boldsymbol{\nu}, \boldsymbol{\Theta}\}. $$

The structural model may be written as

$$ \boldsymbol{\eta}_i^{(g)} = \boldsymbol{\alpha}^{(g)} + \mathbf{B}^{(g)} \boldsymbol{\eta}_i^{(g)} + \boldsymbol{\zeta}_i^{(g)}, $$

with

$$\boldsymbol{\zeta}_i^{(g)} \sim (0, \boldsymbol{\Psi}^{(g)}).$$

where

- $\boldsymbol{\alpha}^{(g)}$ contains latent intercepts  
- $\mathbf{B}^{(g)}$ contains the structural coefficients  
- $\boldsymbol{\zeta}_i^{(g)}$ is a vector of structural disturbance terms  
- $\boldsymbol{\Psi}^{(g)}$ is their covariance matrix

Structural invariance is then defined as

$$\theta_S^{(g)} = \theta_S \quad \forall g,$$

with

$$\theta_S = \{\mathbf{B}, \boldsymbol{\alpha}, \boldsymbol{\Psi}\}. $$

More generally, invariance may be expressed as

$$ \theta^{(g)} = \theta + \Delta^{(g)}, $$

with

$$ \Delta^{(g)} = 0 \quad \forall g. $$

# A.6 Parameter Stability

A parameter is considered empirically stable if

$$\left|\hat\theta_j^{*(1)} - \hat\theta_j^{*(2)}\right| \le \epsilon_j.$$ where $\epsilon_j$ denotes a substantively justified tolerance for parameter $j$.

Parameter stability across the set of theory-relevant parameters may be represented by

$$ \hat{S} = \left( \mathbb{I}\left(\left|\hat\theta_j^{*(1)} - \hat\theta_j^{*(2)}\right| \le \epsilon_j\right) \right)_{j=1}^r, $$ where $\mathbb{I}(\cdot)$ denotes the indicator function.

# A.7 Theoretical Stability

Let

$$C(\theta)$$

denote the theoretical claim implied by a parameter configuration $\theta$. Direct replication may then be interpreted as preservation of the same theoretical claim across independent datasets:

$$C\left(\theta^{*(1)}\right) = C\left(\theta^{*(2)}\right).$$
Conceptual replication is defined as preservation of a theoretical claim across theoretically related estimands:

$$C\left(\theta^{(1)}\right) = C\left(\phi\left(\theta^{(2)}\right)\right),$$

where
$$\phi:\Theta_2 \rightarrow \Theta_1$$

denotes a mapping between different operationalizations.
